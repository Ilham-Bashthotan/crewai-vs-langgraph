# CrewAI Hierarchical Process

```python
"""
CrewAI Hierarchical Pattern - Contoh Sederhana
Menunjukkan cara kerja manager agent yang mengatur worker agents
"""

from crewai import Agent, Task, Crew, Process, LLM
```

```python
# ============================================================================
# 1. SETUP LLM
# ============================================================================
manager_llm = LLM(
    model="ollama/llama3.1:8b",
    base_url="http://localhost:11434",
    temperature=0
)

llm = LLM(
    model="ollama/qwen3:0.6b-q4_K_M",
    base_url="http://localhost:11434",
    temperature=0.7
)
```

```python
researcher = Agent(
    role='Researcher',
    goal='Cari data akurat',
    backstory='Ahli riset data.',
    llm=llm,
    allow_delegation=False 
)

writer = Agent(
    role='Writer',
    goal='Tulis artikel menarik',
    backstory='Penulis konten kreatif.',
    llm=llm,
    allow_delegation=False
)
```

```python
task1 = Task(
    description='Cari tren AI terbaru 2026', 
    expected_output='List tren'
)
task2 = Task(
    description='Tulis blog post dari tren tersebut', 
    expected_output='Artikel blog'
)
```

```python
crew = Crew(
    agents=[researcher, writer],
    tasks=[task1, task2],
    
    process=Process.hierarchical,
    manager_llm=manager_llm,
    tracing=True,
    
    verbose=True
)

result = crew.kickoff()
```
