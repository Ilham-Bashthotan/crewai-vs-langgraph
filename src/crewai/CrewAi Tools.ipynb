```python
from crewai import Agent, Task, Crew, Process, LLM
from crewai_tools import TavilySearchTool
from dotenv import load_dotenv
llm = LLM(
    model="ollama/llama3.1:8b",
    base_url="http://localhost:11434",
)
# load env
load_dotenv()
tavily_tool = TavilySearchTool(
    search_depth="basic",
    topic="general"
)
researcher = Agent(
    role="Senior Researcher",
    goal="Uncover groundbreaking information on the topic",
    backstory="You're a seasoned analyst with a knack for deep dives",
    tools=[tavily_tool],
    llm=llm,
    verbose=True
)
writer = Agent(
    role="Report Writer",
    goal="Craft compelling reports from research",
    backstory="You're a skilled writer who turns data into stories",
    llm=llm,
    verbose=True
)
research_task = Task(
    description="Search the web for the latest developments in {topic} as of 2025",
    expected_output="A bullet list of 10 key findings with sources",
    agent=researcher
)
write_task = Task(
    description="Write a 500-word report based on the research findings",
    expected_output="A well-structured markdown report",
    agent=writer,
    context=[research_task]
)
crew = Crew(
    agents=[researcher, writer],
    tasks=[research_task, write_task],
    process=Process.sequential,
    verbose=True,
    llm=llm
)
result = crew.kickoff(inputs={"topic": "Agentic AI trends in 2025"})
print(result)
```
